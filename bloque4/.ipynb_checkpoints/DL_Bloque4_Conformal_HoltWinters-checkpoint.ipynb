# Predicción conformal con Holt-Winters

```python
import sys
import subprocess

try:
    from statsforecast import StatsForecast
    from statsforecast.models import HoltWinters
    from statsforecast.utils import ConformalIntervals
except ModuleNotFoundError:
    subprocess.check_call([
        sys.executable, "-m", "pip", "install", "statsforecast==2.1.1"
    ])
    from statsforecast import StatsForecast
    from statsforecast.models import HoltWinters
    from statsforecast.utils import ConformalIntervals

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

pd.set_option("display.max_columns", 20)
```

## 1. Carga y preparación de datos

```python
ROOT_PATH = "../data/VolumeForecasting/train_OwBvO8W/"

volume = pd.read_csv(ROOT_PATH + "historical_volume.csv")
volume["ds"] = pd.to_datetime(volume["YearMonth"].astype(str), format="%Y%m")
volume["unique_id"] = volume["Agency"] + "__" + volume["SKU"]
volume = volume.rename(columns={"Volume": "y"})
volume = volume.sort_values(["unique_id", "ds"]).reset_index(drop=True)

volume.head()
```

```python
summary = pd.Series({
    "filas": len(volume),
    "agencias": volume["Agency"].nunique(),
    "skus": volume["SKU"].nunique(),
    "series Agency-SKU": volume["unique_id"].nunique(),
    "fecha inicial": volume["ds"].min(),
    "fecha final": volume["ds"].max(),
    "volúmenes cero": int((volume["y"] == 0).sum()),
})
summary
```

```python
# Auditoría: una fila por serie y mes, frecuencia mensual regular.
duplicates = volume.duplicated(["unique_id", "ds"]).sum()
counts = volume.groupby("unique_id").size()

print("Duplicados serie-fecha:", duplicates)
print("Longitud mínima/máxima por serie:", counts.min(), counts.max())

assert duplicates == 0
assert counts.min() == counts.max()
```

## 2. Separación temporal

Reservamos octubre–diciembre de 2017 como test. Todos los modelos y residuos
conformales se calculan con información disponible hasta septiembre.

La calibración conformal utiliza cuatro ventanas históricas de tres meses cada una.
En cada ventana se simula un pronóstico pasado y se conserva su error. Esos errores
determinan cuánto debe abrirse el intervalo alrededor del pronóstico final.

```python
HORIZON = 3
SEASON_LENGTH = 12
N_CALIBRATION_WINDOWS = 4
LEVELS = [80, 95]
TEST_START = pd.Timestamp("2017-10-01")

sf_cols = ["unique_id", "ds", "y"]
train_df = volume.loc[volume["ds"] < TEST_START, sf_cols].copy()
test_df = volume.loc[volume["ds"] >= TEST_START, sf_cols].copy()

print("Observaciones de entrenamiento:", len(train_df))
print("Observaciones de test:", len(test_df))
print("Último mes de entrenamiento:", train_df["ds"].max())
print("Meses de test:", test_df["ds"].drop_duplicates().dt.strftime("%Y-%m").tolist())

assert len(test_df) == volume["unique_id"].nunique() * HORIZON
```

## 3. Holt-Winters con calibración conformal

`ConformalIntervals` calibra los intervalos a partir de errores fuera de muestra en
ventanas históricas. Un `h=3` independiente dentro de la calibración produce errores
específicos para cada horizonte: mes 1, mes 2 y mes 3.

```python
conformal = ConformalIntervals(
    n_windows=N_CALIBRATION_WINDOWS,
    h=HORIZON,
    method="conformal_distribution",
)

holt_winters = HoltWinters(
    season_length=SEASON_LENGTH,
    error_type="A",
    alias="HoltWinters",
    prediction_intervals=conformal,
)

sf = StatsForecast(
    models=[holt_winters],
    freq="MS",
    n_jobs=-1,
)
```

```python
forecast = sf.forecast(
    df=train_df,
    h=HORIZON,
    level=LEVELS,
)

forecast.head()
```

```python
interval_cols = [
    "HoltWinters-lo-95",
    "HoltWinters-lo-80",
    "HoltWinters",
    "HoltWinters-hi-80",
    "HoltWinters-hi-95",
]

# La demanda es no negativa. Subir el límite inferior a cero no puede excluir
# ninguna observación válida y facilita la interpretación operativa.
forecast[interval_cols] = forecast[interval_cols].clip(lower=0)

assert forecast[interval_cols].notna().all().all()
assert (forecast["HoltWinters-lo-95"] <= forecast["HoltWinters-lo-80"]).all()
assert (forecast["HoltWinters-lo-80"] <= forecast["HoltWinters"]).all()
assert (forecast["HoltWinters"] <= forecast["HoltWinters-hi-80"]).all()
assert (forecast["HoltWinters-hi-80"] <= forecast["HoltWinters-hi-95"]).all()

forecast.head()
```

## 4. Tabla por tienda y SKU

Unimos los pronósticos con los identificadores originales y con la demanda observada
del test. Esta tabla permite consultar el intervalo de cualquier combinación.

```python
id_lookup = volume[["unique_id", "Agency", "SKU"]].drop_duplicates()

results = (
    forecast
    .merge(id_lookup, on="unique_id", how="left", validate="many_to_one")
    .merge(
        test_df.rename(columns={"y": "actual"}),
        on=["unique_id", "ds"],
        how="left",
        validate="one_to_one",
    )
    .sort_values(["Agency", "SKU", "ds"])
    .reset_index(drop=True)
)

results["horizon"] = results.groupby("unique_id").cumcount() + 1
results["covered_80"] = results["actual"].between(
    results["HoltWinters-lo-80"], results["HoltWinters-hi-80"]
)
results["covered_95"] = results["actual"].between(
    results["HoltWinters-lo-95"], results["HoltWinters-hi-95"]
)
results["width_80"] = results["HoltWinters-hi-80"] - results["HoltWinters-lo-80"]
results["width_95"] = results["HoltWinters-hi-95"] - results["HoltWinters-lo-95"]

display_cols = [
    "Agency", "SKU", "ds", "horizon", "actual", "HoltWinters",
    "HoltWinters-lo-80", "HoltWinters-hi-80",
    "HoltWinters-lo-95", "HoltWinters-hi-95",
]
results[display_cols].head(9).round(2)
```

## 5. Evaluación: precisión, cobertura y amplitud

Un intervalo útil debe equilibrar dos objetivos:

1. **Cobertura:** qué proporción de valores reales cae dentro del intervalo.
2. **Amplitud:** qué tan ancho es el intervalo.

Un intervalo enorme puede alcanzar 100% de cobertura y aun así no ser útil para
inventario. Por eso siempre reportamos ambas cantidades.

```python
def evaluate_intervals(frame):
    error = frame["HoltWinters"] - frame["actual"]
    denominator = np.maximum(frame["actual"].abs().sum(), 1e-8)

    return pd.Series({
        "n": len(frame),
        "MAE": error.abs().mean(),
        "RMSE": np.sqrt(np.mean(error ** 2)),
        "WAPE_%": 100 * error.abs().sum() / denominator,
        "Bias": error.mean(),
        "coverage_80_%": 100 * frame["covered_80"].mean(),
        "nominal_80_%": 80.0,
        "mean_width_80": frame["width_80"].mean(),
        "coverage_95_%": 100 * frame["covered_95"].mean(),
        "nominal_95_%": 95.0,
        "mean_width_95": frame["width_95"].mean(),
    })

overall_metrics = evaluate_intervals(results).to_frame("test completo")
overall_metrics.round(2)
```

```python
by_horizon = results.groupby("horizon", sort=True).apply(
    evaluate_intervals,
    include_groups=False,
)
by_horizon.round(2)
```

## 6. Diagnóstico por agencia y SKU

```python
series_diagnostics = (
    results.groupby(["Agency", "SKU"], observed=True)
    .agg(
        mean_actual=("actual", "mean"),
        mae=("actual", lambda x: np.nan),  # se reemplaza abajo con error alineado
        coverage_80=("covered_80", "mean"),
        coverage_95=("covered_95", "mean"),
        mean_width_80=("width_80", "mean"),
        mean_width_95=("width_95", "mean"),
    )
)

series_mae = (
    results.assign(abs_error=(results["HoltWinters"] - results["actual"]).abs())
    .groupby(["Agency", "SKU"])["abs_error"]
    .mean()
)
series_diagnostics["mae"] = series_mae
series_diagnostics[["coverage_80", "coverage_95"]] *= 100
series_diagnostics = series_diagnostics.sort_values("mae", ascending=False)

series_diagnostics.head(10).round(2)
```

## 7. Visualización para una tienda y SKU

La banda oscura representa el intervalo 80%; la clara, el intervalo 95%.

```python
def plot_agency_sku(agency, sku, history_months=24):
    history = volume.loc[
        (volume["Agency"] == agency)
        & (volume["SKU"] == sku)
        & (volume["ds"] < TEST_START)
    ].tail(history_months)

    pred = results.loc[
        (results["Agency"] == agency)
        & (results["SKU"] == sku)
    ].sort_values("ds")

    if history.empty or pred.empty:
        raise ValueError(f"No existe la combinación {agency} / {sku}")

    fig, ax = plt.subplots(figsize=(11, 5))
    ax.plot(history["ds"], history["y"], color="C0", label="Historia")
    ax.plot(pred["ds"], pred["actual"], "o-", color="black", label="Real")
    ax.plot(pred["ds"], pred["HoltWinters"], "o-", color="C1", label="Pronóstico")

    ax.fill_between(
        pred["ds"], pred["HoltWinters-lo-95"], pred["HoltWinters-hi-95"],
        color="C1", alpha=0.15, label="Intervalo 95%",
    )
    ax.fill_between(
        pred["ds"], pred["HoltWinters-lo-80"], pred["HoltWinters-hi-80"],
        color="C1", alpha=0.30, label="Intervalo 80%",
    )
    ax.axvline(TEST_START, color="gray", linestyle="--", alpha=0.7)
    ax.set(
        title=f"Holt-Winters conformal — {agency}, {sku}",
        xlabel="Mes",
        ylabel="Volumen",
    )
    ax.grid(alpha=0.25)
    ax.legend()
    plt.show()

plot_agency_sku("Agency_01", "SKU_01")
```

```python
# Cambia estos identificadores para consultar otra serie.
SELECTED_AGENCY = "Agency_22"
SELECTED_SKU = "SKU_01"

results.loc[
    (results["Agency"] == SELECTED_AGENCY)
    & (results["SKU"] == SELECTED_SKU),
    display_cols,
].round(2)
```
