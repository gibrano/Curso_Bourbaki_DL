# Regularización en Redes Neuronales con TensorFlow

## Cargar dataset

```

import tensorflow as tf
from tensorflow.keras import layers, models

(x_train, y_train), (x_test, y_test) = tf.keras.datasets.fashion_mnist.load_data()
x_train = x_train / 255.0
x_test = x_test / 255.0
```

## Modelo baseline

```

def build_baseline():
    model = models.Sequential([
        layers.Flatten(),
        layers.Dense(256, activation='relu'),
        layers.Dense(128, activation='relu'),
        layers.Dense(10, activation='softmax')
    ])
    model.compile(optimizer='adam', loss='sparse_categorical_crossentropy',
                  metrics=['accuracy'])
    return model

baseline = build_baseline()
history_base = baseline.fit(x_train, y_train, epochs=5, validation_split=0.2)
```

## Modelo con L2

```

l2 = tf.keras.regularizers.l2(0.001)

def build_l2():
    model = models.Sequential([
        layers.Flatten(),
        layers.Dense(256, activation='relu', kernel_regularizer=l2),
        layers.Dense(128, activation='relu', kernel_regularizer=l2),
        layers.Dense(10, activation='softmax')
    ])
    model.compile(optimizer='adam', loss='sparse_categorical_crossentropy',
                  metrics=['accuracy'])
    return model

model_l2 = build_l2()
history_l2 = model_l2.fit(x_train, y_train, epochs=10, validation_split=0.2)
```

## Comparación gráfica

```

import matplotlib.pyplot as plt

plt.plot(history_base.history['val_accuracy'], label='Baseline')
plt.plot(history_l2.history['val_accuracy'], label='L2')
plt.legend()
plt.show()
```
